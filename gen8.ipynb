```python
import os
os.environ["COHERE_API_KEY"] = "nIaoFXDzL5pVzLXtAsSBAnEJfZaAsufQL0IvWEvk"

from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate
from langchain_cohere import ChatCohere

# Read file
with open("climate.txt", "r") as file:
    document = file.read()

# Create Prompt Template
template = """
You are an AI assistant.
Read the following document and provide output in this format:
Topic:
Key Points:
1.
2.
3.

Summary (2 lines):
Document:
{text}
"""

prompt = ChatPromptTemplate.from_messages(
    [
        HumanMessagePromptTemplate.from_template(template)
    ]
)

# Load Cohere chat model
llm = ChatCohere(model="command-r-08-2024", temperature=0.5)

# Create chain
chain = prompt | llm

# Run
print("Sending request...")
result = chain.invoke({"text": document})

print("Generated Output:\n")
print(result.content)
```
