```python
from transformers import pipeline

summarizer = pipeline("text2text-generation", model="t5-small")

text = """
summarize: India is a country located in South Asia and is known for its rich cultural diversity,
historical monuments, traditions, and languages. It is the seventh-largest country in the world
by land area and has one of the largest populations globally. India is famous for landmarks such
as the Taj Mahal, its festivals like Diwali and Holi, and its contributions to science,
technology, and literature. The country has a rapidly growing economy and is one of the major
players in the global IT industry. Agriculture, manufacturing, and services are important sectors
of the Indian economy. India is also known for its democratic system of government and its unity
in diversity.
"""

result = summarizer(text, max_length=50, min_length=15)

print(result[0]['generated_text'])
```
