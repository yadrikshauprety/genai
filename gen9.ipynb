```python
# STEP 1: Import Libraries
import wikipediaapi, requests, re
from bs4 import BeautifulSoup

# STEP 2: Create Wikipedia Object
wiki = wikipediaapi.Wikipedia(
    user_agent="Project",
    language="en"
)

# STEP 3: Create Function
def fetch_info(name):

    # Get Wikipedia page
    page = wiki.page(name)
    html = requests.get(
        f"https://en.wikipedia.org/wiki/{name.replace(' ','_')}"
    ).text
    soup = BeautifulSoup(html, "html.parser")


    # Default values
    founder = "Not Found"
    employees = "Not Found"

    # STEP 4: Extract Information

    year_match = re.search(r"\b(19|20)\d{2}\b", page.text)
    founded = year_match.group() if year_match else "Not Found"

    # Find founder, employees and motto
    for row in soup.select(".infobox tr"):

        header = row.find("th")
        value = row.find("td")

        if header and value:

            key = header.text.lower()

            if any(word in key for word in ["founder", "chairman"]):
                founder = value.get_text(" ", strip=True)

            if any(word in key for word in ["employees", "staff"]):
                employees = value.get_text(" ", strip=True)

    # Find branches
    branches = [
        city for city in
        ["Bangalore", "London", "Tokyo", "New York"]
        if city.lower() in page.text.lower()
    ]

    # STEP 5: Print Result
    print("\nInstitution:", name)
    print("Founder:", founder)
    print("Founded:", founded)
    print("Employees:", employees)
    print("Branches:", ", ".join(branches) if branches else "Not Found")
    print("Summary:", page.summary[:600])


# Input
name = input("Enter Institution Name: ")

# Function Call
fetch_info(name)
```
