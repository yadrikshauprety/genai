```python
# pip install gensim matplotlib scikit-learn

import gensim.downloader as api
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Load pre-trained Word2Vec model
model = api.load("word2vec-google-news-300")

# Technology domain words
words = ["computer", "internet", "software", "hardware", "server",
         "network", "database", "keyboard", "mouse", "programming"]

# Reduce embeddings to 2D using PCA
vectors = [model[w] for w in words]
points = PCA(n_components=2).fit_transform(vectors)

# Display 5 similar words
print("Top 5 words similar to 'computer':")
for word, score in model.most_similar("computer", topn=5):
    print(word, ":", round(score, 4))

# Plot embeddings
for i, word in enumerate(words):
    plt.scatter(points[i,0], points[i,1])
    plt.text(points[i,0], points[i,1], word)

plt.title("Technology Word Embeddings (PCA)")
plt.show()
```
